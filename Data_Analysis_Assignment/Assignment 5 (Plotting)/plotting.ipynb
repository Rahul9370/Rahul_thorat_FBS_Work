```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Refer to the tips.csv dataset. The dataset contains total 244 tips(data about customers
# served by a restaurant) across 7 different columns.
```

```python
df = pd.read_csv("../../dataset/tips.csv")
df
```

```python
# 1. Plot a day wise boxplot of totbill column.
sns.boxplot(x='day', y='total_bill', data=df)
plt.show()
```

```python
# 2. Plot a pariplot by setting hue parameter t
sns.pairplot(data=df, hue="sex")  #sns.pairplot() (from Seaborn) is used for visualizing relationships between multiple numerical variables in a dataset.
plt.show()
```

```python
# 3. Draw a scatter plot of tip and totalbill and also allot the points according to the sex
sns.scatterplot(data= df, x="total_bill", y="tip", hue="sex")
plt.title("Total Bill vs Tip by Sex")
plt.show()
```

```python
# 4. Plot a hitsogram of totbill for 50 bins
sns.histplot(data=df, x='total_bill', bins=50, kde=True)
plt.show()
```

```python
# 5. Plot a heatmap to represent entire dataset's correlation
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', linewidths=0.5)
```

```python
dataset = pd.read_csv('../../dataset/cars93.csv')
```

```python
dataset
```

```python
# 1. Bar Chart:
# a. Create a bar chart to compare the average price of cars by manufacturer.
# a. Label the axes and add a suitable title.
# b. Use different colors for different manufacturers.
```

```python
dataset.head()
```

```python
avg_price = dataset.groupby("Manufacturer")['Price'].mean()
avg_price
```

```python
colors = ['red', 'blue', 'green', 'orange', 'purple', 'black', 'yellow', 'violet']
plt.bar(avg_price.index, avg_price.values, color=colors)
plt.xticks(rotation=90)
plt.xlabel("Manufacturer")
plt.ylabel("Average Price")
plt.title("Average Price of Cars by Manufacturer")
plt.show()
```

```python
# 2. Line Chart:
# a. Create a line chart showing the trend of horsepower for different types of cars.
# b. Include proper legends, labels, and a title
```

```python
# Group by Type and take average Horsepower
hp_trend = dataset.groupby('Type')['Horsepower'].mean()
# Plot line chart
plt.plot(hp_trend.index, hp_trend.values, marker='o', label='Avg Horsepower')

plt.xlabel("Car Type")
plt.ylabel("Average Horsepower")
plt.title("Trend of Horsepower for Different Car Types")

plt.legend()
plt.show()
```

```python
# 3. Pie Chart:
# a. Create a pie chart showing the proportion of different car types in the dataset.
# b. Include percentages in the chart.
```

```python
 # Count number of cars per type
type_counts = dataset['Type'].value_counts()
# Plot pie chart
plt.pie(type_counts.values, labels=type_counts.index, autopct='%1.1f%%', explode=[0,0,0,0,0,0.3])
# Title
plt.title("Proportion of Different Car Types")
plt.show()
```

```python
# 4. Scatter Plot:
# a. Create a scatter plot to show the relationship between horsepower and MPG
# (Miles Per Gallon).
# b. Label the axes and add a title.
```

```python
# Scatter plot
plt.scatter(dataset['Horsepower'], dataset['MPG.city'])
# Labels
plt.xlabel("Horsepower")
plt.ylabel("MPG (Miles Per Gallon)")
# Title
plt.title("Horsepower vs MPG")

plt.show()
```
