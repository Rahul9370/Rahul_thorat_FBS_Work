```python
import numpy as np
import pandas as pd
```

```python
# Part A: Data Preparation
# 1.	Load dataset & display first 5 rows.
# 2.	Check for missing values, datatypes, and descriptive statistics.
# 3.	Categorize employees into age groups: 22–30, 31–40, 41–50, 51+.
```

```python
# 1.	Load dataset & display first 5 rows.
df = pd.read_csv("../dataset/employee_performance_dataset.csv")
df.head()
```

```python
# 2.	Check for missing values, datatypes, and descriptive statistics.
df.isna().sum()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
# 3. Categorize employees into age groups: 22–30, 31–40, 41–50, 51+.
df['Age_Group'] = pd.cut(df['Age'], bins=[21,30,40,50,100],
                         labels=['22-30','31-40','41-50','51+'])
df.head()
```

```python
# Part B: Exploratory Data Analysis (EDA)
# 1.	Find average performance by department.
# 2.	Find the top 10 employees by performance score.
# 3.	Compare average salary across departments.
# 4.	Compare performance by gender.
# 5.	Find correlation between Attendance, Feedback, Trainings, Salary, and Performance.
```

```python
# 1.	Find average performance by department.
df.groupby('Department')['Performance_Score'].mean()
```

```python
# 2.	Find the top 10 employees by performance score.
df.sort_values('Performance_Score', ascending=False).head(10)
```

```python
# 3.	Compare average salary across departments.
df.groupby('Department')['Salary(LPA)'].mean()
```

```python
# 4.	Compare performance by gender.
df.groupby("Gender")['Performance_Score'].mean()
```

```python
# 5.	Find correlation between Attendance, Feedback, Trainings, Salary, and Performance.
df[['Trainings_Attended', 'Customer_Feedback', 'Trainings_Attended', 'Salary(LPA)', 'Performance_Score']].corr()
```

```python
# Part C: Visualization (Matplotlib + Seaborn)
# 1.	Bar chart → Department-wise average performance.
# 2.	Boxplot → Salary distribution by department.
# 3.	Histogram → Age distribution of employees.
# 4.	Heatmap → Correlation matrix.
# 5.	Scatter plot → Salary vs Performance_Score.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1.	Bar chart → Department-wise average performance.
sns.barplot(x="Department", y="Performance_Score", data=df)
plt.title("Department-wise Average Performance")
plt.xticks(rotation=45)
plt.show()
```

```python
# 2.	Boxplot → Salary distribution by department.
sns.boxplot(x="Department", y="Salary(LPA)", data=df)

plt.title("Salary Distribution by Department")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3.	Histogram → Age distribution of employees.
sns.histplot(df["Age"], bins=10, kde=True)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(x="Salary(LPA)", y="Performance_Score", data=df)

plt.title("Salary vs Performance Score")
plt.xlabel("Salary (LPA)")
plt.ylabel("Performance Score")
plt.show()
```
